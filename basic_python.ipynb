Problem E4: Data Manipulation with Lambda Functions

Use a lambda function to:
Create a new column tip_percentage: (tip / total_bill × 100)
Use a lambda function on a categorical column to:
Change the values of time to:
"Afternoon" if Lunch
"Evening" if Dinner
Display the first 5 rows after the transformation.

```python
df["tip_percentage"] = df.apply(
    lambda x: x["tip"] / x["total_bill"] * 100, axis =1)

df
```

```python
student = [["Rina", [90, 85, 80]], ["Tono", [70, 75, 60]], ["Sari", [88, 89, 92]]]

```

Find students whose average score is above 80

```python
for data in student:
    name = data[0]
    score = data[1]
    avg = sum(score)/len(score)

    if avg > 80:
        print(f"{name} passed with an average score of {avg:.2f}")
```

Display students who have at least one score below 70

```python
for data in student:
    if any(n < 70 for n in data[1]):
        print(data[0])
```

From the student data, display the name of the student with the highest average score

```python
max_avg = 0

for data in student:
    name = data[0]
    score = data[1]
    avg = sum(score) / len(score)
    if avg > max_avg:
        max_avg = avg 

print(f"{name} has the highest average score: {max_avg:.2f}")
```
