**Judul**: Validasi Formulir Data Peserta

**Deskripsi**:
Kamu diberikan list of dictionary, di mana setiap dictionary merepresentasikan data peserta. Setiap peserta harus memiliki key 'nama', 'usia', dan 'email'.

Tuliskan program untuk:

1. Menampilkan jumlah total peserta.

2. Mengecek berapa peserta yang memiliki data lengkap (semua key ada).

3. Menampilkan daftar nama peserta yang datanya belum lengkap.

```python
# Data
peserta = [
    {'nama': 'Aldi', 'usia': 17, 'email': 'aldi@gmail.com'},
    {'nama': 'Bela', 'usia': 18},
    {'usia': 19, 'email': 'cici@gmail.com'},
    {'nama': 'Dodi', 'usia': 20, 'email': 'dodi@gmail.com'},
]

# Output:
# Total peserta: 4
# Peserta dengan data lengkap: 2
# Peserta tidak lengkap: ['Bela', 'Unknown']
```

```python
print(f'Total peserta : {len(peserta)}')
```

```python
peserta_lengkap = 0
tidak_lengkap = []

for i in peserta:
    if 'nama' in i and 'usia' in i and 'email' in i:
        peserta_lengkap += 1
    else :
        nama = i.get('nama','unknown')
        tidak_lengkap.append(nama)

print(f"Peserta dengan peserta lengkap : {peserta_lengkap}")
print(f"Peserta tidak lengkap : {tidak_lengkap}")
```

```python
score = [
    {"mapel": "Matematika", "nilai": 80},
    {"mapel": "Bahasa", "nilai": 75},
    {"mapel": "Matematika", "nilai": 90},
    {"mapel": "Bahasa", "nilai": 65},
    {"mapel": "IPA", "nilai": 85},
]

#Buat dictionary yang menampung rata-rata nilai per mapel
```

```python
total_nilai = {}
jumlah_data = {}

for i in score:
    mapel = i['mapel']
    nilai = i['nilai']
    if mapel in jumlah_data:
        jumlah_data[mapel] += 1 
        total_nilai[mapel] += nilai
    else :
        jumlah_data[mapel] = 1
        total_nilai[mapel] = nilai 

rata2 = {}

for mapel in total_nilai:
    rata2[mapel] = total_nilai[mapel] / jumlah_data[mapel]

rata2

```

Buat fungsi `average_rating(data)` → mengembalikan rata-rata dari semua rating gabungan seluruh produk.

```python
def total_penjualan(data):
    total = 0
    for i in data:
        penjualan = i['penjualan']
        total += penjualan 

    return total


total_penjualan([
    {"nama": "Laptop A", "penjualan": 120, "rating": [4.5, 4.0, 5.0]},
    {"nama": "Laptop B", "penjualan": 75, "rating": [4.2, 4.0]},
    {"nama": "Laptop C", "penjualan": 50, "rating": [3.5, 4.0, 3.8, 4.1]}
])
```

```python
def average_rating(data):
    summary_rating = []
    for i in data:
        rating = i['rating']
        summary_rating.append(sum(rating)/len(rating))  
    return sum(summary_rating) / len(summary_rating)

    



average_rating([
    {"nama": "Laptop A", "penjualan": 120, "rating": [4.5, 4.0, 5.0]},
    {"nama": "Laptop B", "penjualan": 75, "rating": [4.2, 4.0]},
    {"nama": "Laptop C", "penjualan": 50, "rating": [3.5, 4.0, 3.8, 4.1]}
])
```

Buat fungsi `rating_tertinggi(data)` yang:

1. Mengembalikan rating tertinggi dari masing-masing produk.

2. Jika rating kosong atau salah format, tangani dengan try-except.

Gunakan built-in:

* `max()` → ambil nilai tertinggi dari list rating

```python
produk = [
    {"nama": "Produk A", "rating": [4.5, 4.7, 5.0]},
    {"nama": "Produk B", "rating": []},
    {"nama": "Produk C", "rating": [4.1, 4.2]}
]

def rating_tertinggi(produk):
    for i in produk:
        try:
            tertinggi = max(i['rating'])
            print(f'Produk {i['nama']} rating tertingginya : {tertinggi}')
        except ValueError as e :
            print(f'Produk {i['nama']} error {e}')

rating_tertinggi(produk)
```
